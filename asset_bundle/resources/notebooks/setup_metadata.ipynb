```sql
%python
from pyspark.sql import *
CATALOG_="bakehouse_dev"
spark.sql(f"create catalog if not exists {CATALOG_}")
spark.sql(f"create schema if not exists {CATALOG_}.p_metadata")
spark.sql(f"create schema if not exists {CATALOG_}.brz")
spark.sql(f"create schema if not exists {CATALOG_}.slv")
spark.sql(f"create schema if not exists {CATALOG_}.gld")
```

```sql
%python
spark.sql(f"""
CREATE TABLE if not EXISTS {CATALOG_}.p_metadata.table_info
(
    table_id int, ---pk
    table_name string,
    table_key string,
    t_source string,
    t_source_path string,
    t_layer string ,
    t_load_type string,
    t_updated_at TIMESTAMP
)
""")


spark.sql(f"""

CREATE or REPLACE TABLE  {CATALOG_}.p_metadata.batch_control (
    batch_id int, ---pk
    batch_start_date TIMESTAMP,
    batch_end_date TIMESTAMP,
    batch_status string,
    batch_error string
)
""")

spark.sql(f"""

CREATE or REPLACE TABLE  {CATALOG_}.p_metadata.dq_checks (
    batch_id string, 
    source_table string,
    target_table string,
    source_rowcount int,
    target_rowcount int,
    diff int,
    null_pk_count int,
    duplicate_pk_count int,
    msg string,
    date timestamp
)
""")

```
